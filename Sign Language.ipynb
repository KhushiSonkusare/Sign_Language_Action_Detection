```python
import numpy as np
import cv2
import mediapipe as mp
from keras.models import load_model


# Load the trained model
model = load_model('C:\\Users\\khush\\OneDrive\\Desktop\\SignLang\\models\\action.keras')  # Update model path if necessary

# Set up Mediapipe for holistic model
mp_holistic = mp.solutions.holistic
mp_drawing = mp.solutions.drawing_utils

def mediapipe_detection(image, holistic):
    """Process the image through MediaPipe holistic model."""
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    image.flags.writeable = False
    results = holistic.process(image)
    image.flags.writeable = True
    image = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)
    return image, results

def extract_keypoints(results):
    """Extract keypoints from MediaPipe results."""
    pose = np.array([[res.x, res.y, res.z, res.visibility] for res in results.pose_landmarks.landmark]).flatten() if results.pose_landmarks else np.zeros(33 * 4)
    face = np.array([[res.x, res.y, res.z] for res in results.face_landmarks.landmark]).flatten() if results.face_landmarks else np.zeros(468 * 3)
    lh = np.array([[res.x, res.y, res.z] for res in results.left_hand_landmarks.landmark]).flatten() if results.left_hand_landmarks else np.zeros(21 * 3)
    rh = np.array([[res.x, res.y, res.z] for res in results.right_hand_landmarks.landmark]).flatten() if results.right_hand_landmarks else np.zeros(21 * 3)
    keypoints = np.concatenate([pose, face, lh, rh])
    return keypoints

# Actions
actions = ['please', 'good', 'game', 'help', 'you', 'play', 'guitar']

# Prepare for real-time prediction (rest of your code)
cap = cv2.VideoCapture(0)
sequence = []
sentence = []
predictions = []
threshold = 0.5
cooldown_frames = 10  # Number of frames to wait between predictions
cooldown_counter = cooldown_frames  # Initialize cooldown counter
ready_timer = 0  # Timer to wait for the next action
ready_duration = 150  # Time to wait (5 seconds at 30 fps)

with mp_holistic.Holistic(min_detection_confidence=0.5, min_tracking_confidence=0.5) as holistic:
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        # Make detections
        image, results = mediapipe_detection(frame, holistic)

        # Draw landmarks (optional, uncomment if needed)
        mp_drawing.draw_landmarks(image, results.pose_landmarks, mp_holistic.POSE_CONNECTIONS)
        mp_drawing.draw_landmarks(image, results.face_landmarks, mp_holistic.FACEMESH_CONTOURS)
        mp_drawing.draw_landmarks(image, results.left_hand_landmarks, mp_holistic.HAND_CONNECTIONS)
        mp_drawing.draw_landmarks(image, results.right_hand_landmarks, mp_holistic.HAND_CONNECTIONS)

        # Extract keypoints
        keypoints = extract_keypoints(results)
        if keypoints.size > 0:  # Proceed only if keypoints are detected
            sequence.append(keypoints)
            sequence = sequence[-30:]  # Keep the last 30 frames

            # Only predict after the ready timer has elapsed
            if ready_timer >= ready_duration:
                if len(sequence) == 30 and cooldown_counter >= cooldown_frames:
                    # Ensure sequence shape matches (1, 30, 1662)
                    sequence_resized = np.expand_dims(sequence, axis=0)  # Shape: (1, 30, 1662)
                    
                    # Make predictions
                    res = model.predict(sequence_resized)[0]
                    predictions.append(np.argmax(res))

                    # Prediction logic with cooldown and exclusion of 'games'
                    if res[np.argmax(res)] > threshold:
                        predicted_word = actions[np.argmax(res)]  # Get the predicted word
                        if predicted_word != 'game' and predicted_word not in sentence:  # Exclude 'game'
                            if len(sentence) > 0:
                                if predicted_word != sentence[-1]:  # If the word is not the same as last
                                    sentence.append(predicted_word)
                                    cooldown_counter = 0  # Reset cooldown
                            else:
                                sentence.append(predicted_word)
                                cooldown_counter = 0  # Reset cooldown

            # Increment ready timer
            ready_timer += 1

        # Increment cooldown counter if still in cooldown period
        if cooldown_counter < cooldown_frames:
            cooldown_counter += 1

        # Display the predicted sentence on the screen
        cv2.rectangle(image, (0, 0), (640, 40), (245, 117, 16), -1)
        cv2.putText(image, ' '.join(sentence), (3, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2, cv2.LINE_AA)

        # Show the image
        cv2.imshow('OpenCV Feed', image)

        # Check if 'q' is pressed to stop the loop and break out
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

cap.release()
cv2.destroyAllWindows()

# Display the full sentence after 'q' is pressed
print('Predicted Sentence:', ' '.join(sentence))

processed_sentence = send_to_groq_model(sentence)

if processed_sentence:
    print(f"Processed Sentence: {processed_sentence}")
    speak(processed_sentence)  # Speak the processed sentence
else:
    print("Failed to process the sentence.")
```
